```python
from ultralytics import YOLO

# Load  YOLOv8n model
model = YOLO("yolov8n.pt")
```

```python
model.train(
    data="/content/drive/MyDrive/final_model/dataSet/data.yaml",    # path to your YAML
    epochs=100,          # number of epochs
    batch=16,            # batch size
    imgsz=416,           # image size
    project="/content/drive/MyDrive/final_model/model_result",  # save results to Drive
    name="project",  # run folder name
    patience=20          # early stopping
)
```

```python
# Load your trained YOLOv8n model (replace with your actual weights path)
model = YOLO("/content/drive/MyDrive/final_model/model_result/project-2/weights/best.pt")
```

```python
# Path to validation images
val_path = "/content/drive/MyDrive/final_model/dataSet/valid/images/"

# Run predictions
results = model.predict(
    source=val_path,  # folder of validation images
    imgsz=416,        # image size used in training
    conf=0.25,        # confidence threshold
    save=True,        # save images with predicted bounding boxes
    save_dir="/content/drive/MyDrive/final_model/predictions"  # folder to save results
)
```

```python
from ultralytics import YOLO
import cv2
from gtts import gTTS
from IPython.display import Audio
import matplotlib.pyplot as plt

# Load your trained model
# model = YOLO("/content/drive/MyDrive/EDBP/yolo_training/blind_project/weights/best.pt")

# Image path (Colab path)
image_path = "/content/drive/MyDrive/EDBP/yolo_training/tree.jpg"

# Run detection
results = model(image_path)

detected_objects = set()

for r in results:
    for box in r.boxes:
        class_id = int(box.cls[0])
        class_name = model.names[class_id]
        confidence = float(box.conf[0])
        detected_objects.add(class_name)

        print(f"Detected: {class_name} | Confidence: {confidence:.2f}")

annotated_frame = results[0].plot()   # Draw boxes

# Convert BGR to RGB for matplotlib
annotated_rgb = cv2.cvtColor(annotated_frame, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8,8))
plt.imshow(annotated_rgb)
plt.axis("off")
plt.title("Detection Result")
plt.show()

# -------------------------
# 🔊 TEXT TO SPEECH SECTION
# -------------------------

if detected_objects:
    speech_text = "I detected " + ", ".join(detected_objects)
else:
    speech_text = "No objects detected"

print("Speaking:", speech_text)

tts = gTTS(text=speech_text, lang='en')
tts.save("output.mp3")

Audio("output.mp3")
```
